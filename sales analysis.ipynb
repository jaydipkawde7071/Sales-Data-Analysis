```python
import pandas as pd

data = {
    "Date": [
        "2026-01-01", "2026-01-05", "2026-01-10",
        "2026-02-01", "2026-02-05", "2026-02-10",
        "2026-03-01", "2026-03-05", "2026-03-10",
        "2026-04-01"
    ],
    "Product": [
        "Laptop", "Mouse", "Keyboard", "Monitor", "Laptop",
        "Chair", "Desk", "Mouse", "Keyboard", "Monitor"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Furniture",
        "Furniture", "Electronics", "Electronics", "Electronics"
    ],
    "Quantity": [2, 10, 5, 3, 1, 4, 2, 15, 8, 5],
    "Price": [55000, 500, 1200, 12000, 55000, 6500, 9000, 500, 1200, 12000],
    "Region": [
        "North", "South", "East", "West", "South",
        "North", "East", "West", "South", "North"
    ]
}

df = pd.DataFrame(data)

df
```

```python
df["Total_Sales"] = df["Quantity"] * df["Price"]

df
```

```python
total_sales = df["Total_Sales"].sum()

print("Total Sales: ₹", total_sales)
```

```python
product_sales = df.groupby("Product")["Total_Sales"].sum()

print(product_sales)
```

```python
import matplotlib.pyplot as plt

product_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby("Region")["Total_Sales"].sum()

print("Region-wise Sales:")
print(region_sales)
```

```python
import matplotlib.pyplot as plt

region_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Region-wise Sales Analysis")
plt.xlabel("Region")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
best_product = product_sales.idxmax()
highest_sales = product_sales.max()

print("Best Selling Product:", best_product)
print("Highest Product Sales: ₹", highest_sales)
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Month"] = df["Date"].dt.month_name()

monthly_sales = df.groupby("Month")["Total_Sales"].sum()

print("Monthly Sales Analysis:")
print(monthly_sales)
```

```python
monthly_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Monthly Sales Analysis")
plt.xlabel("Month")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
best_region = region_sales.idxmax()
highest_region_sales = region_sales.max()

print("Best Performing Region:", best_region)
print("Region Sales: ₹", highest_region_sales)
```
