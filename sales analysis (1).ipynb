```python
import pandas as pd

data = {
    "Date": [
        "2026-01-01", "2026-01-05", "2026-01-10",
        "2026-02-01", "2026-02-05", "2026-02-10",
        "2026-03-01", "2026-03-05", "2026-03-10",
        "2026-04-01"
    ],
    "Product": [
        "Laptop", "Mouse", "Keyboard", "Monitor", "Laptop",
        "Chair", "Desk", "Mouse", "Keyboard", "Monitor"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Furniture",
        "Furniture", "Electronics", "Electronics", "Electronics"
    ],
    "Quantity": [2, 10, 5, 3, 1, 4, 2, 15, 8, 5],
    "Price": [55000, 500, 1200, 12000, 55000, 6500, 9000, 500, 1200, 12000],
    "Region": [
        "North", "South", "East", "West", "South",
        "North", "East", "West", "South", "North"
    ]
}

df = pd.DataFrame(data)

df
```

```python
df["Total_Sales"] = df["Quantity"] * df["Price"]

df
```

```python
total_sales = df["Total_Sales"].sum()

print("Total Sales: ₹", total_sales)
```

```python
product_sales = df.groupby("Product")["Total_Sales"].sum()

print(product_sales)
```

```python
import matplotlib.pyplot as plt

product_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby("Region")["Total_Sales"].sum()

print("Region-wise Sales:")
print(region_sales)
```

```python
import matplotlib.pyplot as plt

region_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Region-wise Sales Analysis")
plt.xlabel("Region")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
best_product = product_sales.idxmax()
highest_sales = product_sales.max()

print("Best Selling Product:", best_product)
print("Highest Product Sales: ₹", highest_sales)
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Month"] = df["Date"].dt.month_name()

monthly_sales = df.groupby("Month")["Total_Sales"].sum()

print("Monthly Sales Analysis:")
print(monthly_sales)
```

```python
monthly_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Monthly Sales Analysis")
plt.xlabel("Month")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
best_region = region_sales.idxmax()
highest_region_sales = region_sales.max()

print("Best Performing Region:", best_region)
print("Region Sales: ₹", highest_region_sales)
```

```python
#python
import pandas as pd
import matplotlib.pyplot as plt

# Calculate Total Sales
df["Total_Sales"] = df["Quantity"] * df["Price"]

# Calculate KPIs
total_sales = df["Total_Sales"].sum()
total_orders = len(df)
total_quantity = df["Quantity"].sum()
best_product = df.groupby("Product")["Total_Sales"].sum().idxmax()

# Display Dashboard Summary
print("=" * 40)
print("       SALES ANALYSIS DASHBOARD")
print("=" * 40)

print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Total Orders: {total_orders}")
print(f"Total Quantity Sold: {total_quantity}")
print(f"Best-Selling Product: {best_product}")

print("=" * 40)

```

```python
#python
product_sales = df.groupby("Product")["Total_Sales"].sum()
product_sales = product_sales.sort_values(ascending=False)

plt.figure(figsize=(10, 5))

product_sales.plot(kind="bar")

plt.title("Product-wise Sales Analysis")
plt.xlabel("Product")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
#python
region_sales = df.groupby("Region")["Total_Sales"].sum()

plt.figure(figsize=(8, 5))

region_sales.plot(kind="bar")

plt.title("Region-wise Sales Analysis")
plt.xlabel("Region")
plt.ylabel("Total Sales (₹)")
plt.tight_layout()
plt.show()

```

```python
#python
# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

# Extract month
df["Month"] = df["Date"].dt.month_name()

# Calculate monthly sales in calendar order
monthly_sales = (
    df.groupby(df["Date"].dt.to_period("M"))["Total_Sales"].sum()
)

# Plot monthly sales
plt.figure(figsize=(10, 5))
monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales (₹)")
plt.xticks(
    range(len(monthly_sales)),
    monthly_sales.index.strftime("%b %Y"),
    rotation=45
)
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
region_sales = df.groupby("Region")["Total_Sales"].sum()

plt.figure(figsize=(7, 7))

plt.pie(
    region_sales,
    labels=region_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Region-wise Sales Distribution")
plt.tight_layout()
plt.show()

```

```python

# Best-selling product
product_sales = df.groupby("Product")["Total_Sales"].sum()
best_product = product_sales.idxmax()

# Highest revenue region
region_sales = df.groupby("Region")["Total_Sales"].sum()
best_region = region_sales.idxmax()

# Highest sales month
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total_Sales"].sum()
best_month = monthly_sales.idxmax()

print("===== BUSINESS INSIGHTS =====")
print(f"Total Revenue: ₹{df['Total_Sales'].sum():,.2f}")
print(f"Best-Selling Product: {best_product}")
print(f"Highest Revenue Region: {best_region}")
print(f"Highest Sales Month: {best_month.strftime('%B %Y')}")
print("=============================")

```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Calculate sales
df["Total_Sales"] = df["Quantity"] * df["Price"]
df["Date"] = pd.to_datetime(df["Date"])

# Prepare data
product_sales = df.groupby("Product")["Total_Sales"].sum().sort_values()
region_sales = df.groupby("Region")["Total_Sales"].sum()
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total_Sales"].sum()

# Dashboard
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Product-wise sales
product_sales.plot(kind="barh", ax=axes[0, 0])
axes[0, 0].set_title("Product-wise Sales")
axes[0, 0].set_xlabel("Sales (₹)")

# Region-wise sales
region_sales.plot(kind="bar", ax=axes[0, 1])
axes[0, 1].set_title("Region-wise Sales")
axes[0, 1].set_ylabel("Sales (₹)")

# Monthly sales trend
monthly_sales.plot(kind="line", marker="o", ax=axes[1, 0])
axes[1, 0].set_title("Monthly Sales Trend")
axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("Sales (₹)")

# Region sales distribution
axes[1, 1].pie(
    region_sales,
    labels=region_sales.index,
    autopct="%1.1f%%",
    startangle=90
)
axes[1, 1].set_title("Region-wise Sales Distribution")

plt.suptitle("SALES ANALYSIS DASHBOARD", fontsize=18)
plt.tight_layout()
plt.show()

# KPI summary
print(f"Total Revenue: ₹{df['Total_Sales'].sum():,.2f}")
print(f"Total Orders: {len(df)}")
print(f"Total Quantity Sold: {df['Quantity'].sum()}")
print(f"Best-Selling Product: {df.groupby('Product')['Total_Sales'].sum().idxmax()}")

```
